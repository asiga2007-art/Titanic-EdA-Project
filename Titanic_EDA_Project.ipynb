# Titanic – Exploratory Data Analysis (EDA)

**Project 01**

This notebook explores the Titanic training dataset using Pandas. It inspects the data, checks missing values and duplicates, cleans selected columns, and studies survival patterns.

**Dataset:** Titanic training data (891 passenger records).

## 1. Import libraries and load the dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/AchilleasDeli/titanic-dataset/main/train.csv"
df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
df.head()
```

## 2. Understand the dataset

```python
# Dataset structure and column data types
df.info()

# Statistical summary of numerical columns
df.describe()
```

## 3. Check missing values and duplicates

```python
print("Missing values in each column:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

## 4. Clean the data

```python
# Fill missing Age values with the median
df["Age"] = df["Age"].fillna(df["Age"].median())

# Fill missing Embarked values with the most frequent value (mode)
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Cabin has many missing values, so remove this column for this analysis
df = df.drop(columns=["Cabin"])

# Remove duplicate rows, if any
df = df.drop_duplicates()

print("Missing values after cleaning:")
print(df.isnull().sum())
print("\nCleaned dataset shape:", df.shape)
```

## 5. Grouping and aggregation

```python
# Survival rate by gender
survival_by_sex = df.groupby("Sex")["Survived"].mean() * 100
print("Survival rate by gender (%):")
print(survival_by_sex.round(1))

# Survival rate by passenger class
survival_by_class = df.groupby("Pclass")["Survived"].mean() * 100
print("\nSurvival rate by passenger class (%):")
print(survival_by_class.round(1))

# Survival rate by port of embarkation
survival_by_port = df.groupby("Embarked")["Survived"].mean() * 100
print("\nSurvival rate by port (%):")
print(survival_by_port.round(1))

# Other aggregations
print("\nMean age and fare by passenger class:")
display(df.groupby("Pclass").agg({"Age": "mean", "Fare": "mean"}).round(2))
```

## 6. Visualize survival patterns

```python
sns.countplot(data=df, x="Sex", hue="Survived")
plt.title("Titanic Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of passengers")
plt.show()

sns.barplot(data=df, x="Pclass", y="Survived")
plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger class")
plt.ylabel("Survival rate")
plt.show()
```
